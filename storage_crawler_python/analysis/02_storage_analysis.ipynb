# 02 — Storage Analysis

Explore storage by drive, hierarchy, extension, category, owner, and modification age.

```python
from pathlib import Path
import pandas as pd

DATA = Path(r"C:\path\to\STORAGE_ANALYSIS.csv")
df = pd.read_csv(DATA, low_memory=False)
files = df[df['ItemType'].eq('File')].copy()
```

```python
# Largest file categories
files.groupby('FileCategory')['SizeGB'].sum().sort_values(ascending=False).head(20)
```

```python
# Largest individual files
files.nlargest(20, 'SizeGB')[['FullPath', 'SizeGB', 'FileCategory', 'Owner']]
```

```python
# Simulation storage
sim = files[files['IsSimulationFile'].eq('Yes')]
sim.groupby('FileCategory')['SizeGB'].sum().sort_values(ascending=False)
```
