# 02 — Storage Analysis

Use DuckDB for aggregation and pandas only for the small result sets returned by SQL.


```python
from pathlib import Path
import duckdb

DB = Path.home() / "Documents" / "crawler_result" / "STORAGE_ANALYSIS.duckdb"
con = duckdb.connect(str(DB), read_only=True)
```

```python
# Storage by category
con.sql("SELECT * FROM category_summary ORDER BY storage_gb DESC").df()
```

```python
# Largest 50 files
con.sql("SELECT * FROM largest_files LIMIT 50").df()
```

```python
# Storage by owner
con.sql("""
SELECT *
FROM owner_summary
ORDER BY storage_gb DESC
LIMIT 50
""").df()
```

```python
# Top hierarchy/category/age combinations
con.sql("""
SELECT *
FROM folder_summary
ORDER BY storage_gb DESC
LIMIT 100
""").df()
```
