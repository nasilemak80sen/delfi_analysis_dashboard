# 03 — Simulation Storage Analysis

Focus on simulation-related storage, age buckets, and stale high-value files.


```python
from pathlib import Path
import duckdb

DB = Path.home() / "Documents" / "crawler_result" / "STORAGE_ANALYSIS.duckdb"
con = duckdb.connect(str(DB), read_only=True)
```

```python
con.sql("""
SELECT *
FROM simulation_summary
ORDER BY storage_gb DESC
""").df()
```

```python
con.sql("""
SELECT *
FROM stale_files
ORDER BY SizeGB DESC
LIMIT 100
""").df()
```

```python
con.sql("""
SELECT
    age_bucket,
    SUM(storage_gb) AS storage_gb,
    SUM(file_count) AS file_count
FROM simulation_summary
GROUP BY age_bucket
ORDER BY storage_gb DESC
""").df()
```
