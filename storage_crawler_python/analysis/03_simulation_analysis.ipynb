# 03 — Simulation Data Analysis

Focused analysis of simulation-related file storage.

```python
from pathlib import Path
import pandas as pd

DATA = Path(r"C:\path\to\STORAGE_ANALYSIS.csv")
df = pd.read_csv(DATA, low_memory=False)
sim = df[df['IsSimulationFile'].eq('Yes')].copy()
sim['LastModified'] = pd.to_datetime(sim['LastModified'], errors='coerce')
sim['AgeDays'] = (pd.Timestamp.now() - sim['LastModified']).dt.days
sim.head()
```

```python
sim.groupby('FileCategory')['SizeGB'].agg(['count', 'sum']).sort_values('sum', ascending=False)
```

```python
# Potentially stale simulation files (>365 days old)
stale = sim[sim['AgeDays'] > 365]
stale[['FullPath', 'SizeGB', 'FileCategory', 'LastModified']].sort_values('SizeGB', ascending=False).head(50)
```
