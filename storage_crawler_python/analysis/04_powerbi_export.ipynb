# 04 — Power BI Dataset Preparation

Create an analytical Parquet dataset and compact summary tables for Power BI.

```python
from pathlib import Path
import pandas as pd

DATA = Path(r"C:\path\to\STORAGE_ANALYSIS.csv")
OUT = DATA.with_suffix('.parquet')
df = pd.read_csv(DATA, low_memory=False)
df.to_parquet(OUT, index=False)
print(OUT)
```

```python
# Example Power BI summary: storage by category
files = df[df['ItemType'].eq('File')]
summary = (
    files.groupby(['Drive', 'FileCategory'], dropna=False)['SizeGB']
    .sum()
    .reset_index()
)
summary
```
