# 04 — Power BI Export

Export only compact analytical tables first. Keep the detailed Parquet inventory outside the report model until we have a reason to expose it.


```python
from pathlib import Path
import duckdb

DB = Path.home() / "Documents" / "crawler_result" / "STORAGE_ANALYSIS.duckdb"
OUT = DB.parent / "analytics"
OUT.mkdir(parents=True, exist_ok=True)
con = duckdb.connect(str(DB), read_only=True)
```

```python
tables = [
    "data_quality",
    "folder_summary",
    "owner_summary",
    "category_summary",
    "simulation_summary",
    "stale_files",
    "largest_files",
]

for table in tables:
    target = OUT / f"{table}.csv"
    con.execute(
        f"COPY {table} TO '{target.as_posix()}' (FORMAT CSV, HEADER TRUE)"
    )

print(f"Exported {len(tables)} tables to {OUT}")
```
